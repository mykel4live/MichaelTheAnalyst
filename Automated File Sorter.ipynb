Automated File Sorter

```python
import os, shutil
```

```python
path = r"C:/Users/44786/Desktop/Data Analysis/Python/Python Tutorial 3/Automated File Sorter/"
```

```python
file_name = os.listdir(path)

print(file_name)
```

```python
os.path.exists(path + 'csv files')
```

```python
folder_names = ['csv files', 'image files', 'text files']

for loop in range(0,3):
    if not os.path.exists(path + folder_names[loop]):
        print(path + folder_names[loop])
        os.makedirs(path + folder_names[loop])
    
```

```python
for file in file_name:
    if ".csv" in file and not os.path.exists(path + "csv files/" +file):
        shutil.move(path + file, path + "csv files/" + file)
    elif ".png" in file and not os.path.exists(path + "images files/" +file):
        shutil.move(path + file, path + "images files/" + file)
    elif ".txt" in file and not os.path.exists(path + "text files/" +file):
        shutil.move(path + file, path + "text files/" + file)
    elif ".xlsx" in file and not os.path.exists(path + "csv files/" +file):
        shutil.move(path + file, path + "csv files/" + file)
    else:
        print("There are files in this path that were not moved")
```
